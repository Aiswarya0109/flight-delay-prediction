## Task 1

Create a function `sorted_random()` which generates a list of *n* random numbers, sorted in descending order (i.e. largest first). The `number *n* should be an input argument for the function. If no value is specified for *n*, by default a list of 10 numbers should be generated.

```python
# first we need to import the random number generation module
import random
```

```python
# define our new function
def sorted_random(n=10):
    numbers = []
    for i in range(n):
        # generate the next random number and add it to our list
        numbers.append(random.random())
    # sort the numbers in descending order
    numbers.sort(reverse=True)
    return numbers
```

```python
# test our function
print(sorted_random(2))
print(sorted_random(4))
# test with the default argument value
print(sorted_random())
```

## Task 2

Download the comma-separated file *scores.csv* from the module Brightspace page and save it to the same directory as your notebook. 

Write a function `display_numbers()` which takes one argument - an input file path. The function should read floating point numbers from each line in the specified file, and compute the sum for the values on each line. Print each sum to 2 decimal places. 

Apply the function `display_numbers()` to *scores.csv*.

```python
def display_numbers(in_path):
    # open the file and process each line
    with open(in_path, "r") as fin:
        for line in fin:
            line = line.strip()
            parts = line.split(",")
            values = []
            # convert each part to a float
            for s in parts:
                values.append(float(s))
            # print the sum of the values with 2 decimal places
            print(f"{sum(values):.2f}")
```

```python
display_numbers("scores.csv")
```

Here's an improved version that includes proper error handling:

```python
def display_numbers_enhanced(in_path):
    try:
        with open(in_path, "r") as fin:
            for line_num, line in enumerate(fin, 1):
                try:
                    # Convert each value to float and calculate sum
                    values = [float(x) for x in line.strip().split(",")]
                    total = sum(values)
                    print(f"{total:.2f}")
                except ValueError as e:
                    print(f"Error on line {line_num}: Invalid number format - {e}")
                except Exception as e:
                    print(f"Error processing line {line_num}: {e}")
    except FileNotFoundError:
        print(f"Error: File '{in_path}' not found")
    except PermissionError:
        print(f"Error: Permission denied accessing '{in_path}'")
    except Exception as e:
        print(f"Unexpected error: {e}")
```

```python
display_numbers_enhanced("scores.csv")
```

## Task 3

Write a function `reverse_numbers()` which takes two arguments - 1) an input file path; 2) an output file path. The function should read floating point numbers from each line in the specified input file. The order of the values from in line should then be reversed, and these lines should be written to the specifed output file.

Apply the function `reverse_numbers()` to the file *scores.csv* from Task 2 above to create a new output file *reversed.csv*.

```python
def reverse_numbers(in_path, out_path):
    # open input and output files
    with open(in_path, "r") as fin, open(out_path, "w") as fout:
        for line in fin:
            # read and clean each line
            line = line.strip()
            parts = line.split(",")
            # reverse the list
            parts.reverse()
            # write the reversed values, separated by commas
            fout.write(",".join(parts))
            # add newline character
            fout.write("\n")
            
    print(f"Successfully reversed numbers from '{in_path}' to '{out_path}'")
```

```python
reverse_numbers("scores.csv", "reversed.csv")
```

Here's an improved version of the function that includes proper error handling:

```python
def reverse_numbers_enhanced(in_path, out_path):
    try:
        with open(in_path, "r") as fin:
            with open(out_path, "w") as fout:
                for l in fin.readlines():
                    l = l.strip()
                    parts = l.split(",")
                    parts.reverse()
                    fout.write(",".join(parts))
                    fout.write("\n")
        print(f"Successfully reversed numbers from '{in_path}' to '{out_path}'")
    except FileNotFoundError:
        print(f"Error: One of the files was not found. Please check the paths: '{in_path}' or '{out_path}'")
    except IOError as e:
        print(f"Error: An I/O error occurred during file operation: {e}")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
```

```python
reverse_numbers_enhanced("scores.csv", "reversed2.csv")
```

## Task 4

Download the comma-separated file *worldcup.csv* from the module Brightspace page and save it to the same directory as your notebook. This data file contains a list of the winners and runners-up for all FIFA World Cup finals, one final per line.

Write a function `count_finalists()` which takes a single argument (an input file path), reads the data, and displays the number of times each country has appeared in a World Cup final. Countries should be displayed in alphabetical order.

Apply the function `count_finalists()` to the file *worldcup.csv*.

```python
def count_finalists(in_path):
    # read the data from the file
    with open(in_path, "r") as fin:
        lines = fin.readlines()
    # skip the header line and process the rest
    counts = {}
    for line in lines[1:]:
        parts = line.strip().split(",")
        winner = parts[1]
        runner_up = parts[2]
        if not winner in counts:
            counts[winner] = 0
        counts[winner] += 1
        if not runner_up in counts:
            counts[runner_up] = 0
        counts[runner_up] += 1
    # sort country names alphabetically
    countries = sorted(counts.keys())
    for country in countries:
        print(f"{counts[country]}\t{country}")
```

```python
count_finalists("worldcup.csv")
```

Note that, if we wanted to sort the countries in descending order by the number of times that they have appeared in a final, we could use 
a Python `Counter` from the `collectons` module, and call its associated `most_common()` function as shown below:

```python
from collections import Counter

def count_finalists_v2(in_path):
    # read the data from the file
    with open(in_path, "r") as fin:
        lines = fin.readlines()
    # skip the header line and process the rest
    counts = Counter()
    for line in lines[1:]:
        parts = line.strip().split(",")
        winner = parts[1]
        runner_up = parts[2]
        counts[winner] += 1
        counts[runner_up] += 1
    # sort countries by their corresponding count
    for country, count in counts.most_common():
        print(f"{count}\t{country}")    
```

```python
count_finalists_v2("worldcup.csv")
```

## Task 5

Write a function `validate_csv_data()` which takes a file path as input and validates that each line contains only numeric values (floats or integers). The function should:

1. Read the file line by line
2. Check if each value in each line can be converted to a float
3. Print any validation errors with the line number
4. Return the total number of valid lines processed

Test your function with the *scores.csv* file and also create a test file with some invalid data to show that the validation process works correctly.

```python
def validate_csv_data(file_path):
    valid_lines = 0
    try:
        # read the file
        with open(file_path, "r") as fin:
            for line_num, line in enumerate(fin, 1):
                line = line.strip()
                if not line:  # skip empty lines
                    continue
                try:
                    values = line.split(",")
                    # try to convert each value to float
                    for i, value in enumerate(values):
                        float(value.strip())
                    # if we get here, all values are valid
                    valid_lines += 1
                    print(f"Line {line_num}: Valid ({len(values)} values)")
                except ValueError as e:
                    print(f"Line {line_num}: Invalid data - '{value.strip()}' is not a number")
                except Exception as e:
                    print(f"Line {line_num}: Error processing line - {e}")
                    
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found")
        return 0
    except Exception as e:
        print(f"Error reading file: {e}")
        return 0
    
    return valid_lines
```
