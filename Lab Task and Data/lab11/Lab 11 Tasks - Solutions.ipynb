# Lab Tasks - Solutions

In this notebook we will apply basic time series analysis to stock price data for Meta and Amazon, sourced from Yahoo Finance.

```python
from datetime import datetime
import pandas as pd
import matplotlib
import matplotlib.pylab as plt
%matplotlib inline 
pd.options.display.float_format = '{:,.2f}'.format
```

## Task 1

Load the stock price CSV files for both Meta (*stock-meta.csv*) and Amazon (*stock-amazon.csv*) into separate Pandas DataFrames.

(Note: Dates in these files are stored in "day/month/year" format)


```python
# load the first CSV file
df_meta = pd.read_csv("stock-meta.csv", index_col="date", parse_dates=True, date_format="%d/%m/%Y")
print(f"Read {df_meta.shape[0]} observations for Meta")
display(df_meta.head())

# load the second CSV file
df_amazon = pd.read_csv("stock-amazon.csv", index_col="date", parse_dates=True, date_format="%d/%m/%Y")
print(f"Read {df_amazon.shape[0]} observations for Amazon")
display(df_amazon.head())
```

Check the time range covered by the data for both stocks.

```python
start_meta, end_meta = df_meta.index.min(), df_meta.index.max()
print(f"Meta date range: {start_meta.date()} to {end_meta.date()}")

start_amazon, end_amazon = df_amazon.index.min(), df_amazon.index.max()
print(f"Amazon date range: {start_amazon.date()} to {end_amazon.date()}")
```

## Task 2

For each of the DataFrames, extract only the **close** stock prices for the year 2021. This should give you two new Pandas Series.

```python
# extract 2021 close prices for Meta
meta_2021 = df_meta.loc["2021"]["close"]
print(f"Extracted {meta_2021.shape[0]} observations from 2021 for Meta")
display(meta_2021.head())

# extract 2021 close prices for Amazon
amazon_2021 = df_amazon.loc["2021"]["close"]
print(f"Extracted {amazon_2021.shape[0]} observations from 2021 for Amazon")
display(amazon_2021.head())
```

## Task 3

Create a new time series DataFrame with two columns, one for the Meta 2021 closing prices and one for the Amazon 2021 closing prices.

```python
# create combined DataFrame with both stocks
df_combined = pd.DataFrame({
    "Meta": meta_2021,
    "Amazon": amazon_2021
})
df_combined.head()
```

## Task 4

Using the data from Task 3, plot the closing prices for Meta and Amazon, using a 7-day rolling mean.

```python
# apply the rolling mean
df_rm = df_combined.rolling(7).mean()
# create the plot
p = df_rm.plot(figsize=(11, 5.5), fontsize=12, zorder=3)
plt.xlabel("Date", fontsize=12)
plt.title("7-Day Rolling Mean of Stock Prices", fontsize=12)
plt.ylabel("Close Price ($)", fontsize=12)
plt.grid(axis="y")
plt.show()
```

## Task 5 

Resample the DataFrame from Task 3 to monthly time windows, and plot the new monthly time series.

```python
# convert from day frequency to month (ME) frequency, by averaging the values
df_monthly = df_combined.resample("ME").mean()

# create the plot
p = df_monthly.plot(figsize=(11, 5.5), fontsize=12, lw=2, zorder=3)
plt.title("Monthly Average Stock Prices", fontsize=12)
plt.ylabel("Closing Stock Price", fontsize=12)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Close Price ($)", fontsize=12)
plt.grid(axis="y")
plt.show()
```
