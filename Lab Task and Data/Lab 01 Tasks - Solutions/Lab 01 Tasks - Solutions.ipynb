The code below creates two Python lists - one list with English words, and a corresponding list with French words.

```python
english = ["hello", "cat", "dog", "yes", "tomorrow", "yesterday", "fish", 
          "difficult", "easy", "bad", "no", "tuesday", "january", "world"]
```

```python
french = ["bonjour", "chat", "chien", "oui", "demain", "hier", "poisson",
          "difficile", "facile", "mauvais", "non", "mardi", "janvier", "monde"]
```

## Task 1

Verify that: 1) both lists contain an equal number of values; 2) all elements in both lists are strings.

(Hint: We can use the `isinstance()` function to check whether a value has a specified type)

```python
# check if the lists have the same length
if len(english) == len(french):
    print("Both lists have the same number of values")
else:
    print("The lists have different lengths")

# check first list contains only strings
for value in english:
    if not isinstance(value, str):
        print("Not a valid string:", value)

# check second list contains only strings
for value in french:
    if not isinstance(value, str):
        print("Not a valid string:", value)        
```

## Task 2

Create a Python dictionary, which maps English words to French words, using the lists above. 

```python
# create an empty dictionary to store English-to-French word pairs
translator = {}
# loop through the English list with index using enumerate
for i, word in enumerate(english):
    # add the English word as the key and the matching French word as the value
    translator[word] = french[i]
```

```python
translator["tuesday"]
```

## Task 3

Remove all key-value pairs from the dictionary above, where the keys (i.e. the English words) contain less than 4 characters.

```python
# create a list of words to remove (avoid modifying dict during iteration)
words_to_remove = []
for word in translator.keys():
    if len(word) < 4:
        words_to_remove.append(word)

# remove the words
for word in words_to_remove:
    del translator[word]
```

```python
translator
```

## Task 4 

For the remaining words, print an alphabetically-ordered list of the English words and their corresponding French translations.

```python
# iterate over a sorted list of English words (dictionary keys)
for word in sorted(translator.keys()):
    # print the English word and its French translation
    print("%s -> %s" % (word, translator[word]))
```

## Task 5

Write a new function named `translate()` which accepts 3 arguments: a translation dictionary, a word to translate, and a default word. If the word is in the dictionary, the function should return the translated word. Otherwise, it should return the default word.

```python
def translate(td, word, default_word):
    # return the translation if the word exists in the dictionary
    if word in td:
        return td[word]
    # otherwise, return the default word
    return default_word
```

```python
translate(translator, "hello", "unknown")
```

```python
translate(translator, "world", "unknown")
```

```python
translate(translator, "december", "unknown")
```

## Task 6

Write a new function named `reverse_translate()` which accepts 2 arguments: a translation dictionary and a word in the target translation language. If the translation is found, it returns the original word; otherwise, it returns a default message indicating that the translation is not available

```python
def reverse_translate(td, translated_word):
    # check if the translated word exists as a value in the dictionary
    for original_word, translation in td.items():
        if translation == translated_word:
            return original_word
    # if not found, return a default message
    return "not available"
```

```python
reverse_translate(translator, "demain")
```

```python
reverse_translate(translator, "bonjour")
```

```python
reverse_translate(translator, "ville")
```
